# Pandas
* Stands for Panel data.
* Built on top of NumPy.
* Revolves around two data structures `Series` which is 1D and `DataFrame` which is 2D.

## What is Pandas used for ?
* Reading and writing data from various file formats like `CSV`, `Excel` and `SQL` databases.
* `Cleaning`, `Merging`, `Joining` and `Reshaping` datasets.
* Performing statistical analysis.
* Visualizing data quickly.

## Pandas Basic Operations
|S No.|Process|Description|
|:---|:---|:---|
|**1**|Data Preprocessing|Handle missing values|
|**2**|Data Cleaning|Remove duplicates and fix inconsistencies|
|**3**|Data Transformation|Filter, sort and modify data|
|**4**|Data Visualization|Represent data graphically|
